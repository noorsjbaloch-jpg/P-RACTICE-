### Exercise 1 — Even or Odd
Create a variable `number = 7`. Print `"Even"` if it is even, otherwise `"Odd"`.
*Hint: a number is even if `number % 2 == 0`.*

```python
# Your code here
number = 7
if number % 2 == 0:
    print(f"{number} is even.")
else:
    print(f"{number} is odd.")
```

### Exercise 2 — Temperature check
Given `temp = 38`, print:
- `"Hot"` if temp is above 30
- `"Warm"` if temp is between 20 and 30 (inclusive)
- `"Cold"` otherwise

```python
# Your code here
temp = 38
if temp >30:
    print("It's hot.")
elif temp > 20 and temp<30:
    print("It's warm.")
else:
    print("It's cold.")
```

### Exercise 3 — Login check
Given `username = "admin"` and `password = "1234"`, print `"Access granted"` only if BOTH the username is `"admin"` AND the password is `"1234"`. Otherwise print `"Access denied"`.

### Exercise 4 — Is it a vowel?
Given `letter = "e"`, check if it is a vowel. 
*Hint: use `if letter in "aeiou":`*

```python
# Your code here
letter = "e"

if letter in "aeiou":
    print("The letter is a vowel.")
else:
    print("The letter is not a vowel.")
```

### Exercise 5 — Shopping cart
Given `cart = []` (empty list), print `"Your cart is empty"` if it has nothing in it, otherwise print how many items it has.
*Hint: an empty list is falsy.*

```python
# Your code here
cart = []
if len(cart) == 0:
    print("Your cart is empty.")
else:
    print(f"Your cart has {len(cart)} items in it.")
```

### Exercise 6 — Traffic light 🚦
Given `light = "yellow"`, print:
- `"Stop"` if red
- `"Go"` if green
- `"Slow down"` if yellow
- `"Invalid light"` for anything else

```python
# Your code here
light = "yellow"
if light == "green":
    print("Go.")
elif light == "yellow":
    print("Slow down.")
elif light == "red":
    print("Stop.")
else:
    print("Invalid light")
```
